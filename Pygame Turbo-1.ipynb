## Funções Lines and circles

O pgzero tem algumas funções embutidas, neste [link](https://pgturbo.readthedocs.io/builtins.html#Screen) podemos ver todas as funções.

Vamos nos concentrar nestas:

Como desenhar linhas e circulos com os comandos:

- screen.draw.circle(( x e Y),r , c)
- screen.draw.filled_circle(( x e Y),r , c)
  - onde x e y são as cordenadas do centro
  - r é o raio
  - c é a cor

- screen.draw.line((x1,y1),(x2,y2), c)
  - Aqui x1 e y1 - ponto inicial
  - x2 e y2 - ponto final da linha;
  - c = cor

- A função draw é chamada pelo ** Pygame Zero** para reenderizar a tela numa taxa de aproximadamente 60 quadros por segundo.
